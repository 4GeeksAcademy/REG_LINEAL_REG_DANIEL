```python
# Convierto las str a numemricas

df['COUNTY_NAME'] = pd.factorize(df['COUNTY_NAME'])[0]
df['STATE_NAME'] = pd.factorize(df['STATE_NAME'])[0]

df.info()
```

# modelo

Pirmero correremos el modelos, ya que Lasso lleva la pendiente a cero de las variables que no tienen correlacion, lo que me permite descartar gran cantidad de ellas.

```python
# Normalizamos

from sklearn.preprocessing import StandardScaler

scaler_norm = StandardScaler()
df_norm = pd.DataFrame(scaler_norm.fit_transform(df),index = df.index,columns= df.columns)
```

```python
from sklearn.model_selection import train_test_split

x = df_norm.drop('Obesity_number',axis = 1)
y = df_norm['Obesity_number']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Corremos el modelo

from sklearn.linear_model import Lasso

lasso_model = Lasso(alpha=0.1,max_iter= 1000)

lasso_model.fit(x_train,y_train)
y_pred = lasso_model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error

r2 = r2_score(y_pred, y_test)
error2 = root_mean_squared_error(y_pred,y_test)

print(f'el r2 es {r2} y el error cuadratico medio es {error2}')
```

# ANALISIS

- El R2 de 0.97 es un muy buen modelo, aunque tambien podria indiqcar que hay variables correlacionadas con la objetivo
- El error cuadratico es bajo


# Visualizamos que variables sobrevivieron a Lasso

```python
import pandas as pd
import matplotlib.pyplot as plt

# Crear un DataFrame con los nombres de las columnas y sus coeficientes
coeficientes = pd.DataFrame({
    'Variable': x_train.columns,
    'Coeficiente': lasso_model.coef_
})

# Filtrar solo las variables que el modelo NO eliminó (coeficiente != 0)
variables_seleccionadas = coeficientes[coeficientes['Coeficiente'] != 0].sort_values(by='Coeficiente', ascending=False)

print(f"Variables originales: {len(coeficientes)}")
print(f"Variables que dejó Lasso: {len(variables_seleccionadas)}")
print(variables_seleccionadas)
```

# CONCLUSION

- sobrevivio el 2% de las variables, lo que indica que la gran mayoria llego su pendiente a cero. (no tienen importancia para la variable objetivo)
